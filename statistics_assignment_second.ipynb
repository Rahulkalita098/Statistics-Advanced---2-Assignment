### Question 1

What is hypothesis testing in statistics?

**Answer:** Hypothesis testing is a statistical method used to make decisions or inferences about population parameters based on sample data. It evaluates assumptions and determines whether evidence is strong enough to reject a null hypothesis.

### Question 2

What is the null hypothesis, and how does it differ from the alternative hypothesis?

**Answer:** The null hypothesis (H₀) assumes no effect or no difference. The alternative hypothesis (H₁) suggests there *is* an effect or difference.

### Question 3

Explain the significance level in hypothesis testing.

**Answer:** The significance level (α), usually 0.05, represents the probability of rejecting the null hypothesis when it is actually true (Type I error).

### Question 4

What are Type I and Type II errors?

**Answer:**
- **Type I Error:** Rejecting a true null hypothesis.
- **Type II Error:** Failing to reject a false null hypothesis.

### Question 5

What is the difference between a Z-test and a T-test?

**Answer:** Z-test is used when population variance is known and sample size is large; T-test is used when population variance is unknown and/or sample size is small.

### Question 9

Z-score calculation.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 5, 1000)
z_scores = (data - np.mean(data)) / np.std(data)
plt.hist(z_scores)
plt.show()
```
